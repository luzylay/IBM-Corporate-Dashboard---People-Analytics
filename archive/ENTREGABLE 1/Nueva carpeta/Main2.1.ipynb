**2.ANÁLISIS DETALLE VENTA**

```python
detalle_ventas = pd.read_csv(r"C:\Users\iccni\OneDrive\Documentos\ESTUDIOS\IBM CURSO IA\TRABAJO EN GRUPO - PROYECTO\BASE DE DATOS\Detalle_ventas.csv",encoding="utf-8",sep=";")
print(detalle_ventas)
```

```python
#Descripción base de datos
detalle_ventas.describe()
detalle_ventas.info()
detalle_ventas.value_counts()
```

```python
#Medidas básicas estadísticas
media_detalle = detalle_ventas["importe"].mean()

mediana_detalle = detalle_ventas["importe"].median()

moda_detalle = detalle_ventas["importe"].mode()

desviacion_detalle = detalle_ventas["importe"].std()

maximo_detalle = detalle_ventas["importe"].max()

minimo_detalle = detalle_ventas["importe"].min()

cuartiles_detalle = detalle_ventas["importe"].quantile([0.25,0.5,0.75,1])
                                  
rango_detalle = detalle_ventas["importe"].max() - detalle_ventas["importe"].min()

correlación_detalle = detalle_ventas[["importe","cantidad"]].corr()

print("media =",media_detalle,".",
      "mediana =",mediana_detalle,
      "moda =",moda_detalle,
      "desviación =",desviacion_detalle,
      "maximo =",maximo_detalle,
      "minimo =",minimo_detalle)

print("rango =",rango_detalle)

print("cuartiles =",cuartiles_detalle)

print("correlación =", correlación_detalle)
```

**Distribución = Multimodal**

```python
#GRAFICO BARRAS

#1 configurar la información
df_importe = detalle_ventas["importe"].sort_values()

#2 Crear un lienzo
plt.figure(figsize=(8,4))
#3 graficar
plt.bar(df_importe.index, df_importe.values,color="blue")

#4 título
plt.title("Importe por venta")
plt.xlabel("id_venta")
plt.ylabel("importe")
plt.xlim(0, 10)
plt.ylim(0, 5000)

# 5. elementos varios
plt.grid(True)
plt.tight_layout()

# 6. mostrar y guardar imagen
plt.show()
```

```python
#GRAFICO CAJA Y BIGOTE

# Asumiendo que en data va tu DataFrame
#Se busca comparar la variable numérica 'variable_y'
# Entre diferentes grupos definidos por 'variable_x'

sns.boxplot(x= detalle_ventas["cantidad"], y=detalle_ventas["importe"], color = "orange", showmeans = True ,meanprops = {"marker":"o", "markerfacecolor":"white", "markeredgecolor":"black"})
plt.show()
```

```python
#GRAFICO LINEAS

#1 configurar la información
df_importe = detalle_ventas["importe"].sort_values()
x = detalle_ventas["id_venta"]
y = detalle_ventas["importe"]

#2 Crear un lienzo
plt.figure(figsize=(8,4))
#3 graficar
plt.plot(x,y,linestyle="-",marker="o",color="purple")

#4 título
plt.title("Importe por venta")
plt.xlabel("id_venta")
plt.ylabel("Importe")

# 5. elementos varios
plt.grid(True)
plt.tight_layout()


# 6. mostrar y guardar imagen
plt.show()
```
